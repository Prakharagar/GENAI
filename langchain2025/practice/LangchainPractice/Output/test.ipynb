```python
from typing import TypedDict
from datetime import date

class CUSTOMDICT(TypedDict):
    name: str
    age: int
    salary: float
    dob: date

mydict={
    'name': 'abc',
    'age': 'ww',
    'salary': 10000,
    'dob': '2025-07-25'
        }
obj=CUSTOMDICT(**mydict)
print(type(obj))
```

```python
from pydantic import BaseModel,Field
from datetime import date

class myclass(BaseModel):
    name: str
    age: int = Field(gt=0,le=100,description="age of person")
    salary: float
    dob: date

dict1={
    'name': 'abc',
    'age': 100,
    'salary': 10000,
    'dob': '2025-10-25'
}
obj=myclass(**dict1)
print(obj)
print(type(obj))
print(obj.dob)

print(obj.model_dump_json())
```
